CLASE PERSONA
Define una clase Persona con atributos como nombre, edad y profesión. Luego, crea varios objetos de esta clase y muestra su información.

```python
class Persona:
    def __init__(self,nombre,edad,profesion):
        self.nombre=nombre
        self.edad=edad
        self.profesion=profesion
    def informacion(self):
        print("Nombre:", self.nombre)
        print("Edad:",self.edad)
        print("Profesión:",self.profesion)
jonas = Persona("Jonás",39,"Fontanero y futuro programador.\n")
irene = Persona("Irene",36,"Enfermera\n")
lucas = Persona("Lucas",3,"Darle guerra a susu amados padres\n")
jonas.informacion()
irene.informacion()
lucas.informacion()
```

CALCULADORA BÁSICA
Crea una clase llamada “Calculadora” con métodos para sumar, restar, multiplicar y dividir. Crea objetos de esta clase y realiza algunas operaciones básicas.

```python
class Calculadora:
    def __init__(self,num1,num2):
        self.num1=num1
        self.num2=num2
    def suma(self):
        resultado=self.num1+self.num2
        print(f"El resultado de sumar {self.num1} y {self.num2} es:{resultado}")
    def resta(self):
        resultado= self.num1-self.num2
        print(f"El resultado de restar {self.num1} y {self.num2} :{resultado}")
    def multiplicar(self):
        resultado=self.num1*self.num2
        print(f"El resultado de multiplicar {self.num1} y {self.num2} es:{resultado}")
    def dividir(self):
        resultado=self.num1/self.num2
        print(f"El resultado de dividir {self.num1} y {self.num2} es:{resultado}")
suma=Calculadora(50,10)
suma.suma()
resta = Calculadora(50,10)
resta.resta() 
multiplicacion= Calculadora(50,10)
multiplicacion.multiplicar()
division=Calculadora(50,10)
division.dividir()        

```

LIBRO
Crea una clase “Libro” con atributos como título, autor y año de publicación. Luego, crea varios objetos Libro y muestra su información.

```python
class Libro:
    def __init__(self,titulo,autor,año):
        self.titulo=titulo
        self.autor=autor
        self.año=año
    def mostrar_info(self):
        print(f"El libro {self.titulo} cuyo autor {self.autor} salió en el año {self.año}")
libro1= Libro("Cien años de soledad" ,"Gabriel García Márquez" ,1967)
libro2= Libro("Don Quijote de la Mancha" , "Miguel de Cervantes Saavedra" , 1605)
libro3= Libro(" El amor en los tiempos del cólera"," Gabriel García Márquez", 1985)
libro1.mostrar_info()
libro2.mostrar_info()
libro3.mostrar_info()
```

RECTÁNGULO
Crea una clase “Rectangulo” con atributos de longitud y ancho. Implementa un método para calcular el área y el perímetro del rectángulo.

```python
class Rectangulo:
    def __init__(self,longitud,alto):
        self.longitud=longitud
        self.alto=alto
    def area(self):
        area=self.longitud*self.alto
        print(f"El area del rectangulo es: {area}")
    def perimetro(self):
        perimetro=2*(self.longitud+self.alto)
        print(f"El perimetro del rectangulo es: {perimetro}")
rectangulo1= Rectangulo(50,25)
rectangulo1.area()
rectangulo1.perimetro()
```
