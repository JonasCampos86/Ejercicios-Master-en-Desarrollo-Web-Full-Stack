10.Cambia la forma del array para que tenga una estructura de tipo (filas, columnas)

```python
import numpy as np
longitud = int(input("pon la longitud del array: "))
array_usuario = np.ones(longitud)
print(array_usuario)
filas=int(input("Dime cuantas filas quieres  "))
columnas = int(input("Dime cuantas colunmas  quieres "))

if filas*columnas==longitud:
    array_usuario = array_usuario.reshape(filas,columnas)
    print(array_usuario)
else:
    print("no se puede crear este array")
```

11.Crea una “matriz identidad” con la misma forma que el array anterior (filas, columnas)

```python
import numpy as np
longitud = int(input("pon la longitud del array: "))
array_usuario = np.ones(longitud)
print(array_usuario)
filas=int(input("Dime cuantas filas quieres  "))
columnas = int(input("Dime cuantas colunmas  quieres "))

if filas*columnas==longitud:
    array_usuario = array_usuario.reshape(filas,columnas)
    matrix_identidad = np.eye(filas,columnas)
    print(array_usuario)
    print(matrix_identidad)
else:
    print("no se puede crear este array")
```
